```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from tabulate import tabulate
```

##### Table 1

```python
data = {
    "UE (V)" : [30.0],
    "UB1 (V)" : [14.9],
    "UB2 (V)" : [14.9],
    "UC1 (V)" : [23.3],
    "UC2 (V)" : [23.3]
}

table = pd.DataFrame (data, index=[0])
#table.reset_index(drop = True,inplace = True)
print (tabulate(table, headers="keys", tablefmt="psql"))
```

#### Table 2

```python
data = {
    "UB1 (V)" : [14.9, 15.0, 15.2, 14.7, 14.6], 
    "UC1 (V)" : [23.3, 22.6, 22.0, 23.9, 24.6],
    "UC2 (V)" : [23.3, 23.8, 24.4, 22.7, 22.1]
}

table = pd.DataFrame (data, index=["Table 1", "+0.1", "+0.3", "-0.2", "-0.3"])
print (tabulate(table, headers="keys", tablefmt="psql"))
```

```python
plt.figure (figsize=(8, 6))

plt.subplot (121)
plt.plot (data["UB1 (V)"], data["UC1 (V)"], 'g')
plt.xlabel ("UB1 (V)")
plt.ylabel ("UC1 (V)")
plt.grid (True)

plt.subplot (122)
plt.plot (data["UB1 (V)"], data["UC2 (V)"], 'm')
plt.xlabel ("UB1 (V)")
plt.ylabel ("UC2 (V)")
plt.grid (True)

plt.tight_layout ()

plt.show ()
```
